```python
import random

# Machine Initialization
machines = ['A', 'B', 'C']

# Parameters
N_play = 10
epsilon = 0.2

# Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# Given reward patterns
rewards = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of which reward to use
reward_index = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

# Run for 10 plays
for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Epsilon-Greedy Decision
    if r < epsilon:
        # Exploration
        selected_machine = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        best_machines = [
            machine for machine in machines
            if Q[machine] == max_q
        ]

        selected_machine = random.choice(best_machines)
        decision = "Exploitation"

    # Get reward automatically
    index = reward_index[selected_machine]

    if index < len(rewards[selected_machine]):
        reward = rewards[selected_machine][index]
        reward_index[selected_machine] += 1
    else:
        # If all given rewards are used, repeat the last reward
        reward = rewards[selected_machine][-1]

    # Update N
    N[selected_machine] += 1

    # Update R
    R[selected_machine] += reward

    # Update Q using incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )

    # Update total reward
    total_reward += reward

    # Display play details
    print(f"Play {play}")
    print(f"Random Number: {r:.4f}")
    print(f"Decision: {decision}")
    print(f"Selected Machine: {selected_machine}")
    print(f"Reward: {reward}")
    print(f"N[{selected_machine}] = {N[selected_machine]}")
    print(f"R[{selected_machine}] = {R[selected_machine]}")
    print(f"Q[{selected_machine}] = {Q[selected_machine]:.2f}")
    print("-" * 40)

# Final Results
print("\n========== FINAL RESULTS ==========")
print("Machine\tN\tR\tQ")
print("-----------------------------")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]:.2f}\t"
        f"{Q[machine]:.2f}"
    )

print("-----------------------------")
print(f"Total Reward after {N_play} plays = {total_reward}")
```
